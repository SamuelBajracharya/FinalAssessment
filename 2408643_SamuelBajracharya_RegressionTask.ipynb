#Function for Test Train Split

```python
# spliting function to split the dataset into training and test sets

def train_test_split(x, y, test_size = 0.2, random_seed = 42):

  np.random.seed(random_seed) #takes random using random seed

  indices = np.arange(x.shape[0]) #arranges the x input
  np.random.shuffle(indices) #shuffles the x input for random selection

  test_split_size = int(len(x) * test_size) #finds the size for test data

  test_indices = indices[:test_split_size] #takes 0-test_split_size for test
  train_indices = indices[test_split_size:] # takes test_split_size to n for train

  x_train, x_test = x[train_indices], x[test_indices] #splits the data using the indices and stores then
  y_train, y_test = y[train_indices], y[test_indices]

  return x_train, x_test, y_train, y_test
```

##Spliting the data into test and train set:

```python
# Split the data into training and test sets (80% train, 20% test)
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size=0.2, random_seed=25)
```

#Function for Scaling data using Normalization



```python
def scalar_X_data(X_train, X_test):
  """
    Splits the dataset into training and testing sets.

    Parameters:
    ----------
    x : ndarray
        The feature matrix containing independent variables.

    y : ndarray
        The target variable (dependent variable).

    test_size : float, optional (default=0.2)
        The proportion of the dataset to include in the test split.
        Must be between 0 and 1.

    random_seed : int, optional (default=42)
        The seed for the random number generator to ensure reproducibility.

    Returns:
    -------
    x_train : ndarray
        The training set features.

    x_test : ndarray
        The testing set features.

    y_train : ndarray
        The training set target variable.

    y_test : ndarray
        The testing set target variable.
  """
  # finding min and max of the training matrix
  X_min_train = X_train.min(axis = 0)
  X_max_train = X_train.max(axis = 0)

  # finding min and max of the testing matrix
  X_min_test = X_test.min(axis = 0)
  X_max_test = X_test.max(axis = 0)

  # scaling the data
  X_train_scaled = (X_train - X_min_train) / (X_max_train - X_min_train)
  X_test_scaled = (X_test - X_min_test) / (X_max_test - X_min_test)

  return X_train_scaled, X_test_scaled
```

##Scaling the Data:

```python
# scaling the feature matrix of train and test
X_train_scaled, X_test_scaled = scalar_X_data(X_train, X_test)
```

#Creating a Linear Regression Model

## Y Prediction function:

```python
def predict_y(X, W, b):
  """
    Computes the predicted values (Y_pred) using a linear regression model.

    Parameters:
    ----------
    X : ndarray
        The feature matrix (input data), where each row is a data sample
        and each column is a feature.

    W : ndarray
        The weight vector (coefficients for each feature).

    b : float
        The bias term (intercept).

    Returns:
    -------
    Y_pred : ndarray
        The predicted target values.
  """
  Y_pred = X.dot(W) + b
  return Y_pred
```

## Cost Function:

```python
def cost_function(X, Y, W, b):
    """
    Compute the cost function for linear regression.
    Parameters:
    ----------
    X : ndarray
        The feature matrix (input data) with shape (n_samples, n_features).

    Y : ndarray
        The actual target values (ground truth) with shape (n_samples,).

    W : ndarray
        The weight vector (coefficients for each feature) with shape (n_features,).

    b : float
        The bias term (intercept).

    Returns:
    -------
    cost : float
        The computed cost value.
    """
    n = len(Y)
    Y_pred = predict_y(X, W, b)
    loss = (Y - Y_pred) ** 2
    cost = (1 / (2 * n)) * np.sum(loss)
    return cost
```

## Gradient Descent:

```python
def gradient_descent(X, Y, W, b, alpha, iterations, show_cost = True):
    """
    Perform gradient descent to learn W and b.
    Parameters:
    ----------
    X : ndarray
        The feature matrix with shape (n_samples, n_features).

    Y : ndarray
        The actual target values with shape (n_samples,).

    W : ndarray
        The weight vector (coefficients for each feature) with shape (n_features,).

    b : float
        The bias term (intercept).

    alpha : float
        The learning rate that controls the step size of parameter updates.

    iterations : int
        The number of gradient descent iterations.

    show_cost : bool, optional (default=True)
        Whether to display the cost history at each iteration.
    Returns:
    -------
    W : ndarray
        The optimized weight vector after gradient descent.

    b : float
        The optimized bias term.

    cost_history : ndarray
        A record of the cost function values at each iteration.
    """
    m = len(Y)
    cost_history = np.zeros(iterations)

    for i in range(iterations):
        Y_pred = predict_y(X, W, b)

        # Compute gradients
        dW = (-1 / m) * X.T.dot(Y - Y_pred)
        db = (-1 / m) * np.sum(Y - Y_pred)

        # Update parameters
        W -= alpha * dW
        b -= alpha * db

        # Store cost
        cost_history[i] = cost_function(X, Y, W, b)

    return W, b, cost_history
```

## Linear Regression:

```python
def linear_regression(X_train = X_train_scaled, alpha_value = 0.01, iteration_value = 5000, show_cost = False):
  """
    Performs linear regression using gradient descent to optimize the weight vector (W) and bias (b).

    Parameters:
    ----------
    X_train : ndarray, optional (default=X_train_scaled)
        The training feature matrix with shape (n_samples, n_features).

    alpha_value : float, optional (default=0.01)
        The learning rate for gradient descent, controlling the step size of each update.

    iteration_value : int, optional (default=5000)
        The number of iterations for gradient descent to run.

    show_cost : bool, optional (default=False)
        Whether to display the cost at certain iterations during training.

    Returns:
    -------
    W_optimal : ndarray
        The optimized weight vector after training.

    b_optimal : float
        The optimized bias term after training.

    cost_history : ndarray
        A record of the cost function values at each iteration.
  """
  W = np.zeros(X_train.shape[1])
  b = 0
  alpha = alpha_value
  iterations = iteration_value
  W_optimal, b_optimal, cost_history = gradient_descent(X_train, Y_train, W, b, alpha, iterations)

  for i in range(iterations):
    if show_cost and ((i % 100 == 0) or (i == iterations - 1)):
      print(f"Iteration {i}: Cost = {cost_history[i]:2f}")

  return W_optimal, b_optimal, cost_history
```

# Creating a KNN Regression Model

## Function to find Euclidean Distance

```python
# Function to compute Euclidean distance
def euclidean_distance(point1, point2):
    """
    Computes the Euclidean distance between two points.

    Parameters:
    ----------
    point1 : ndarray
        The first point, represented as a vector (1D array or a list of numerical values).

    point2 : ndarray
        The second point, represented as a vector (1D array or a list of numerical values).

    Returns:
    -------
    float
        The Euclidean distance between the two points.

    Raises:
    ------
    ValueError
        If the dimensions of the two points do not match, a ValueError will be raised.
    """
    if point1.shape != point2.shape:
        raise ValueError("The dimensions of points do not match, therefore cannot calculate distance.")

    return np.sqrt(np.sum((point1 - point2) ** 2))
```

## Function for predicting Single

```python
# Function to predict a single point using KNN Regression
def predict_single(query, X_train, y_train, k=3):
    """
    Predicts the target value for a single query point using K-Nearest Neighbors (KNN) regression.

    Parameters:
    ----------
    query : ndarray
        The query point (a vector of feature values) for which we want to make a prediction.

    X_train : ndarray
        The training feature matrix (n_samples, n_features).

    y_train : ndarray
        The target values corresponding to the training data (n_samples,).

    k : int, optional (default=3)
        The number of nearest neighbors to consider when making the prediction.

    Returns:
    -------
    float
        The predicted target value for the query point, which is the mean of the `k` nearest neighbors' target values.

    Raises:
    ------
    ValueError
        If the query point has a different dimension than the training data.
    """
    distances = [euclidean_distance(x, query) for x in X_train]
    sorted_indices = np.argsort(distances)
    nearest_indices = sorted_indices[:k]
    nearest_values = y_train[nearest_indices]

    return np.mean(nearest_values)  # Use mean for regression
```

## Function to predict KNN

```python
# Function to predict multiple points using KNN Regression
def predict_KNN(X_test, X_train, y_train, k=3):
    """
    Predicts the target values for multiple query points using K-Nearest Neighbors (KNN) regression.

    Parameters:
    ----------
    X_test : ndarray
        The feature matrix for the test data (n_samples, n_features), where each row is a query point.

    X_train : ndarray
        The training feature matrix (n_samples, n_features).

    y_train : ndarray
        The target values corresponding to the training data (n_samples,).

    k : int, optional (default=3)
        The number of nearest neighbors to consider when making each prediction.

    Returns:
    -------
    ndarray
        An array of predicted target values for each query point in `X_test`.
    """
    return np.array([predict_single(query, X_train, y_train, k) for query in X_test])
```

#Evaluation of the models

## Calculate MSE:

```python
def mse(Y, Y_pred):
    """
    Computes the Mean Squared Error (MSE) between the actual and predicted target values.

    Parameters:
    ----------
    Y : ndarray
        The actual target values (n_samples,).

    Y_pred : ndarray
        The predicted target values (n_samples,).

    Returns:
    -------
    float
        The Mean Squared Error between the actual and predicted values.
    """
    return np.mean((Y - Y_pred) ** 2)
```

## Calculate RMSE:

```python
def rmse(Y, Y_pred):
    """
    Computes the Root Mean Squared Error (RMSE) between the actual and predicted target values.

    Parameters:
    ----------
    Y : ndarray
        The actual target values (n_samples,).

    Y_pred : ndarray
        The predicted target values (n_samples,).

    Returns:
    -------
    float
        The Root Mean Squared Error between the actual and predicted values.
    """
    return np.sqrt(np.mean((Y - Y_pred) ** 2))
```

##Calculate R-squared:

```python
def R2(Y, Y_pred):
    """
    Computes the R-squared (R²) score, also known as the coefficient of determination,
    which indicates the proportion of the variance in the dependent variable that is
    predictable from the independent variables.

    Parameters:
    ----------
    Y : ndarray
        The actual target values (n_samples,).

    Y_pred : ndarray
        The predicted target values (n_samples,).

    Returns:
    -------
    float
        The R-squared score, a value between 0 and 1. A higher value indicates a better fit.
    """
    Y_mean = np.mean(Y)
    SSR = np.sum((Y - Y_pred) ** 2)
    SST = np.sum((Y - Y_mean) ** 2)
    return 1 - (SSR / SST)
```

#Hyperparameter Tuning For Linear regression


```python
def Linear_regression_tuning(alpha_values, iterations_values):
  """
    Tunes the hyperparameters (alpha and number of iterations) for linear regression
    using a grid search approach to maximize the R-squared (R²) score.

    The function iterates over different values of learning rate (alpha) and the number
    of iterations to find the combination that results in the best R² score.

    Parameters:
    ----------
    alpha_values : list
        Takes in a list of alpha values to find the best hyperparameter.

    iterations_values : list
        Takes in a list of iterations values to find the best hyperparameter.

    Returns:
    -------
    best_alpha : float
        The optimal learning rate (alpha) for the model.

    best_iteration : int
        The optimal number of iterations for training.

    best_r2 : float
        The highest R-squared score achieved by the model.
  """
  # Initialize variables to track the best R² score and corresponding hyperparameters
  best_r2 = -np.inf
  best_alpha = None
  best_iteration = None

  for i in range(len(alpha_values)):
      alpha = alpha_values[i]
      for j in range(len(iterations_values)):
        iterations = iterations_values[j]

        # Train the model with the current alpha and iteration values
        W_optimal, b_optimal, cost_history = linear_regression(alpha_value = alpha, iteration_value = iterations, show_cost = False)
        Y_pred = predict_y(X_test_scaled, W_optimal, b_optimal)

        # Evaluate performance using R² score
        r2_evaluate = R2(Y_test, Y_pred)

        # Update the best hyperparameters if the current R² is better
        if r2_evaluate > best_r2:
            best_r2 = r2_evaluate
            best_alpha = alpha
            best_iteration = iterations
  return best_alpha, best_iteration, best_r2

# Call the tuning function and print the results
alpha_list = [0.01, 0.02, 0.03, 0.04, 0.5, 0.1]
iterations_list = [1000, 5000, 10000, 20000]
best_alpha, best_iteration, best_r2 = Linear_regression_tuning(alpha_list, iterations_list)
print(f"The best hyperparameters is alpha: {best_alpha}, iterations: {best_iteration}")
print(f"The best R2 value is: {best_r2}")
```

# Hyperparameter Tuning For KNN Regression

```python
def KNN_regression_tuning(k_values):
    """
    Tunes the hyperparameter `k` (number of neighbors) for K-Nearest Neighbors (KNN) regression
    by evaluating the R-squared (R²) score for various values of `k`.

    The function iterates over different `k` values to find the one that results in the highest R² score.

    Parameters:
    ----------
    k_values : list
        Takes in a list of k values to find the best hyperparameter.

    Returns:
    -------
    best_k : int
        The optimal number of neighbors for the KNN model.

    best_r2 : float
        The highest R-squared score achieved by the KNN model.
    """
    best_r2 = -np.inf
    best_k = None

    for i in range(len(k_values)):
        k = k_values[i]
        predicted_values = predict_KNN(X_test_scaled, X_train_scaled, Y_train, k = k)

        r2_evaluate = R2(Y_test, predicted_values)
        if r2_evaluate > best_r2:
            best_r2 = r2_evaluate
            best_k = k

    return best_k, best_r2

k_list = [3, 5, 7, 9, 11, 13, 15, 17]
best_k, best_r2 = KNN_regression_tuning(k_list)
print(f"The best hyperparameters is k: {best_k}")
print(f"The best R2 value is: {best_r2}")
```

#Function for feature selection

```python
from sklearn.feature_selection import SelectKBest, f_regression

# Declare a function to select features
def feature_selection(k_value = 5):
  """
  Performs feature selection using the SelectKBest method with f_regression as the scoring function.

  Parameters:
  ----------
  k_value : int, optional (default=5)
      The number of top features to select based on their relevance to the target variable.

  Returns:
  -------
  feature_select_df : DataFrame
      A Pandas DataFrame containing the feature names and their corresponding R-statistic scores,
      sorted in descending order of importance.

  X_train_selected : ndarray
      The transformed training dataset containing only the selected features.

  X_test_selected : ndarray
      The transformed testing dataset containing only the selected features.
  """
  X = GDI_dataframe.drop(['HDI Rank', 'Country', 'GDI_Value', 'GDI_Group'], axis = 1)
  y = GDI_dataframe['GDI_Value']

  # Initialize SelectKBest with f_regression
  selector = SelectKBest(score_func = f_regression, k = k_value)

  # Fit selector to the data
  selector.fit(X, y)
  selected_feature_indices = np.where(selector.get_support())[0]

  # Transform the training and testing datasets
  X_train_selected = selector.fit_transform(X_train_scaled, Y_train)
  X_test_selected = selector.transform(X_test_scaled)

  # Create a DataFrame of selected features and scores
  feature_names = X.columns
  feature_scores = selector.scores_
  feature_select_df = pd.DataFrame({'Feature': feature_names, 'R-Statistic': feature_scores})

  # Sort features by importance
  feature_select_df = feature_select_df.sort_values(by = 'R-Statistic', ascending = False)

  return feature_select_df, X_train_selected, X_test_selected
```

```python
def final_model():
  # Call the function to find optimal w and b
  W_optimal, b_optimal, cost_history = linear_regression(X_train = X_train_selected, alpha_value = best_alpha, iteration_value = best_iteration, show_cost = True)

  # Make predictions on test set
  Y_pred = predict_y(X_test_selected, W_optimal, b_optimal)

  # Compute mse, rmse and r2 on test set
  mse_value_final = mse(Y_test, Y_pred)
  rmse_value_final = rmse(Y_test, Y_pred)
  r2_value_final = R2(Y_test, Y_pred)

  # Print the results
  print(f"MSE for Final Model(LR) test: {mse_value_final:4f}")
  print(f"RMSE for Final Model(LR) test: {rmse_value_final:4f}")
  print(f"R2 for Final Model(LR) test: {r2_value_final:4f}")

  # Make predictions on train set
  Y_pred_t = predict_y(X_train_selected, W_optimal, b_optimal)

  # Compute r2 on train set
  mse_value_final_t = mse(Y_train, Y_pred_t)
  rmse_value_final_t = rmse(Y_train, Y_pred_t)
  r2_value_final_t = R2(Y_train, Y_pred_t)
  print(f"MSE for Final Model(LR) training: {mse_value_final_t:4f}")
  print(f"RMSE for Final Model(LR) training: {rmse_value_final_t:4f}")
  print(f"R2 for Final Model(LR) training: {r2_value_final_t:4f}")
  return cost_history
```

```python
cost_history = final_model()
```
